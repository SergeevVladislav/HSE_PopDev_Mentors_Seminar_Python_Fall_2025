```python
import sys
from pathlib import Path
from importlib import import_module
from importlib.metadata import version

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Working folder:", Path.cwd())
missing = []
for package in ["numpy", "pandas", "matplotlib", "seaborn", "statsmodels"]:
    try:
        import_module(package)
        print(package, version(package))
    except ImportError:
        missing.append(package)
        print(package, "MISSING")
assert not missing, "Install requirements.txt in this kernel and restart it."
```

```python
import pandas as pd

for filename in ["synthetic_freelists.csv", "synthetic_respondents.csv", "synthetic_piles.csv", "card_krueger_1992.csv", "synthetic_training.csv"]:
    path = Path("data") / filename
    assert path.is_file(), f"Missing {path}; keep the downloaded folder structure."
    frame = pd.read_csv(path)
    print(filename, frame.shape)
print("All five input files are available.")
```
