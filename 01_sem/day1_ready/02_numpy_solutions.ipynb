## NP-01 · Import NumPy and build an array

Import NumPy using the usual alias np. Create a Python list [1000, 1500, 2000],
then convert it to an array named population_start.
Compare multiplying the list by 2 with multiplying the array by 2.
Calculate a 2% increase for every town and inspect the array's shape.
All town populations in this notebook are fictional.


```python
import numpy as np

population = [1000, 1500, 2000]
population_start = np.array(population)
print("List repeated twice:", population * 2)
print("Each array value doubled:", population_start * 2)
print("Populations after 2% growth:", population_start * 1.02)
print("Array shape:", population_start.shape)
```

## NP-02 · Calculate changes and select with a Boolean mask

The same towns have end-of-year populations [1050, 1530, 1980].
Calculate the change in people and the percentage change for each town.
Create a mask for positive percentage changes, use it to select them, and count them.
Also calculate the mean percentage change across the three towns.
The order of towns must be the same in both arrays.


```python
population_end = np.array([1050, 1530, 1980])
changes = population_end - population_start
growth_percent = changes / population_start * 100
positive = growth_percent > 0

print("Change in people:", changes)
print("Change in percent:", growth_percent)
print("Positive changes (%):", growth_percent[positive])
print("Number of growing towns:", positive.sum())
print("Mean town growth (%):", growth_percent.mean())
```

## NP-03 · Ageing and the denominator

The numbers of people aged 65+ in the same towns are [200, 150, 500].
Calculate their percentage shares using population_start as the denominator.
Select the shares that are at least 20%. Then calculate the combined older population share
across all three towns: total older people / total population × 100.
Compare the combined share with the simple mean of town shares.


```python
older_population = np.array([200, 150, 500])
older_shares = older_population / population_start * 100
combined_share = older_population.sum() / population_start.sum() * 100

print("Town shares (%):", older_shares)
print("Town shares at or above 20%:", older_shares[older_shares >= 20])
print(f"Share across all residents: {combined_share:.2f}%")
print(f"Mean of the three town shares: {older_shares.mean():.2f}%")
```

Across the three towns, 850 of 4,500 residents are aged 65+ (18.89%). The simple mean is 18.33% because it gives the three towns equal weight.

## NP-X1 · Two-dimensional arrays and axis

Create a two-dimensional array whose rows are [1000, 200], [1500, 150], and [2000, 500].
Rows are towns; columns are total population and older population. Inspect shape,
select the second column, and calculate column totals with axis=0.


```python
towns = np.array([[1000, 200], [1500, 150], [2000, 500]])
print("Rows and columns:", towns.shape)
print("Older residents in each town:", towns[:, 1])
print("Column totals (all residents, older residents):", towns.sum(axis=0))
```
