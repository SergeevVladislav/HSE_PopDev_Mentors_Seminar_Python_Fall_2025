## PY-02 · Variables, numbers, text, and types

Fictional Town A has 1,000 residents, including 200 people aged 65 or older.
Store the town name and both counts. Calculate the older population share in percent.
Print an explanatory sentence, then inspect the types of the count, the percentage,
and the town name. Change the older population to 250 and recalculate.

Formula: older population / total population × 100. Use meaningful variable names.
`=` assigns a value; Python names are case-sensitive. Use a decimal point, not a comma.


```python
town = "Town A"
population = 1000
older_population = 200
older_share = older_population / population * 100
print(f"{town} has {older_share:.0f}% of its residents aged 65 or older.")
print(type(population), type(older_share), type(town))

older_population = 250
older_share = older_population / population * 100
print(f"With {older_population} older residents, the share is {older_share:.0f}%.")
```

## PY-03 · Lists and dictionaries: positions and names

Create a list called populations with 1,000, 1,500, and 2,000 residents for three fictional towns.
Print the first value, the first two values, and the total population.
Then create a dictionary for Town A with the keys name and population.
Retrieve its name by key.

A list keeps values in order; its first position is 0. A slice excludes its stopping position.
A dictionary connects keys to values. Use square brackets to select, round brackets to call,
and braces to create a dictionary. Select the second town.


```python
populations = [1000, 1500, 2000]
print("First town:", populations[0])
print("First two towns:", populations[:2])
print("All three towns:", sum(populations))
print("Second town:", populations[1])

area = {"name": "Town A", "population": populations[0]}
print(area["name"])
```

## PY-04 · Comparisons and if / else

For a fictional planning exercise, flag an area when its share of people aged 65+ is at least 15%.
Start with a share of 20. Print the result of the comparison, then use if / else to print
either "Review service needs" or "Below the exercise threshold".
Test the boundary value 15 and a value below it, such as 10.

The 15% threshold is invented for this coding exercise; it is not a policy standard.
`==` compares values. `>=` includes equality. Indentation defines a block of code.


```python
for share in [20, 15, 10]:
    print(f"{share}% meets the threshold:", share >= 15)
    if share >= 15:
        print("Review service needs")
    else:
        print("Below the exercise threshold")
```

## PY-05 · A for loop: repeat an instruction

Use the populations list from PY-03. In a fictional scenario, each town grows by 2% in one year.
Write a for loop that prints the projected population for each town.
Explain which line repeats and which variable takes a new value on each iteration.
This is a scenario with an assumed growth rate, not a fitted demographic forecast.


```python
for population in populations:
    projected_population = population * 1.02
    print(f"{population:,} residents grow to {projected_population:,.0f} in this scenario.")
```

On each pass, `population` takes the next value from the list. Both indented lines run once for that town.

## PY-06 · A function: name a reusable calculation

Define a function called percent with inputs part and total. It should return part / total × 100.
Call it for 200 out of 1,000 and for 300 out of 2,000.
Explain the difference between defining a function and calling it.
Assume a strictly positive total for this introductory function.


```python
def percent(part, total):
    return part / total * 100

print(percent(200, 1000))
print(percent(300, 2000))
```

The definition gives Python the calculation to use. Each call supplies a new pair of values and returns its own percentage.

## PY-07 · Explain a population change

A fictional population rises from 100,000 to 100,500.
Store the start and end values, calculate the absolute change, and use your percent function
to calculate percentage change relative to the starting population.
Print one sentence with both numbers and their units.


```python
population_start = 100_000
population_end = 100_500
change = population_end - population_start
growth_percent = percent(change, population_start)
print(f"The population increased by {change} people, or {growth_percent}% of the starting population.")
```

## PY-X2 · Protect a calculation

Create safe_percent: return None when total is zero or negative; otherwise return the percentage.
Test both branches. None means no usable result, not the number zero.


```python
def safe_percent(part, total):
    if total <= 0:
        return None
    return percent(part, total)

print(safe_percent(200, 1000))
print(safe_percent(0, 1000))
print(safe_percent(200, 0))
print(safe_percent(200, -1000))
```
